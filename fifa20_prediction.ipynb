```python
df[['ls_base', 'ls_modifier']] = (
    df['ls']
    .str.split('+', expand=True)
    .astype(float)
)
```

```python
print(
    df[['short_name', 'ls', 'ls_base', 'ls_modifier']].head(10)
)
```

```python
for col in position_cols:

    split_values = df[col].str.split('+', expand=True)

    df[f'{col}_base'] = pd.to_numeric(
        split_values[0], errors='coerce'
    )

    df[f'{col}_modifier'] = pd.to_numeric(
        split_values[1], errors='coerce'
    )
```

```python
print("Dataset shape:", df.shape)

print(
    df[
        ['short_name', 'ls', 'ls_base', 'ls_modifier',
         'st', 'st_base', 'st_modifier']
    ].head(10)
)
```

```python
base_cols = [f'{col}_base' for col in position_cols]
modifier_cols = [f'{col}_modifier' for col in position_cols]

print("Base columns:", len(base_cols))
print("Modifier columns:", len(modifier_cols))

print("\nMissing values in base columns:")
print(df[base_cols].isnull().sum().sum())

print("\nMissing values in modifier columns:")
print(df[modifier_cols].isnull().sum().sum())
```

```python
print("Base rating range:")
print(df[base_cols].describe().loc[['min', 'max']])

print("\nModifier range:")
print(df[modifier_cols].describe().loc[['min', 'max']])
```

```python
ls_reconstructed = (
    df['ls_base'].astype('Int64').astype(str)
    + '+'
    + df['ls_modifier'].astype('Int64').astype(str)
)

valid_ls = df['ls'].notna()

print(
    "All LS values reconstructed correctly:",
    (ls_reconstructed[valid_ls] == df.loc[valid_ls, 'ls']).all()
)
```

```python
for col in position_cols:

    reconstructed = (
        df[f'{col}_base'].astype('Int64').astype(str)
        + '+'
        + df[f'{col}_modifier'].astype('Int64').astype(str)
    )

    valid = df[col].notna()

    result = (
        reconstructed[valid] == df.loc[valid, col]
    ).all()

    print(f"{col}: {result}")
```

```python
strategy_a = df[
    (df['age'] <= 23) &
    (df['overall'] >= 80) &
    (df['potential'] >= 85)
]

strategy_b = df[
    (df['age'] <= 23) &
    (df['overall'] >= 70) &
    (df['potential_gap'] >= 15)
]

print("Ready-to-perform talent:", len(strategy_a))
print("Development-focused talent:", len(strategy_b))
```

```python
overlap = set(strategy_a['sofifa_id']) & set(strategy_b['sofifa_id'])

print("Strategy A players:", len(strategy_a))
print("Strategy B players:", len(strategy_b))
print("Players in both strategies:", len(overlap))

print(
    "Unique players identified:",
    len(set(strategy_a['sofifa_id']) |
        set(strategy_b['sofifa_id']))
)
```
